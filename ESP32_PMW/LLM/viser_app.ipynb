# Bench 实验 · viser 界面

ESP 每次 reset（LED 测试、录像实验）时保持线圈电源关闭，看到 READY 再开电源。
结束后：uv run python controller/control/safe_off.py，关电源。

```python

from pathlib import Path
import sys
PROJECT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p/'LLM/bench.py').exists())
if str(PROJECT) not in sys.path: sys.path.insert(0, str(PROJECT))
from importlib import reload
import LLM.app as app
app = reload(app)

PORT = '/dev/cu.usbserial-0001'  # 本机 ESP 端口，运行前核对
CAMERAS = (1, 0)  # 左 A = 设备 1，右 B = 设备 0
MODE = (1280, 800, 120)  # 宽、高、fps；换模式需重新标定
EXPOSURE_MS = 8.0  # 运行录像的手动曝光（毫秒）；120 fps 一帧 8.33 ms，不能更长
CALIBRATION_EXPOSURE_MS = 10.0  # 仅标定录像用；超过一帧，相机自动降帧，标定不需要高帧率
GAIN = 100  # 仅手动曝光时使用
SESSION_ID = None  # None 新建；继续时填已有 session ID
CALIBRATION = 'latest'  # 'latest' = 该分辨率下最新的已存标定；或 stereo_rig.json 路径；None = 不加载。需重标时在页面点 Calibrate
DUTY_DIST = dict(scale_pct=15.0, floor_pct=30.0)  # 每次运行每个线圈独立随机 duty：100 - Exp(scale)，低于 floor 重抽；记录在 raw/run_parameters.json
```

```python
app.main(PORT, CAMERAS, MODE, exposure_ms=EXPOSURE_MS, gain=GAIN, session=SESSION_ID,
         calibration=CALIBRATION, duty_dist=DUTY_DIST, calibration_exposure_ms=CALIBRATION_EXPOSURE_MS)
# Interrupt 结束；摄像头、串口随之释放。
```
